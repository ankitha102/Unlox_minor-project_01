**Feature 1** — **WhatsApp Chat Parser**

This section reads the WhatsApp chat export and converts each valid message into structured data containing the date, time, sender, and message text.

```python
for line in lines[:10]:
    print(line.strip())
```

```python
messages = []
system_messages = 0
media_messages = 0
deleted_messages = 0

print("Storage created successfully!")
```

```python
sample = lines[0].strip()

print(sample)
print()
print(sample.split(" - ", 1))
```

```python
for line in lines:
    line = line.strip()

    if line == "":
        continue


    parts = line.split(" - ", 1)

    if len(parts) < 2:
        continue

    timestamp = parts[0]
    content = parts[1]

    sender_parts = content.split(": ", 1)


    if len(sender_parts) < 2:
        system_messages += 1
        continue

    sender = sender_parts[0]
    message = sender_parts[1]


    if message == "<Media omitted>":
        media_messages += 1

    if message == "This message was deleted":
        deleted_messages += 1

    messages.append({
        "timestamp": timestamp,
        "sender": sender,
        "message": message
    })

print("Parser completed!")
```

#Feature 4 — NumPy Activity Heatmap

This section uses NumPy to create a 6 × 24 activity matrix showing how frequently each participant sends messages during every hour of the day.

```python
import numpy as np

print("NumPy version:", np.__version__)
```

```python
# FEATURE 4: NUMPY ACTIVITY HEATMAP

import numpy as np

participants_list = sorted(participants)

activity_matrix = np.zeros((len(participants_list), 24), dtype=int)

for msg in messages:
    timestamp = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")

    sender = msg["sender"]
    hour = timestamp.hour

    row = participants_list.index(sender)

    activity_matrix[row][hour] += 1

print("Activity Matrix:")
print(activity_matrix)
```

```python
#  Display Activity Matrix

print("Participant-wise Hourly Activity")
print("-" * 80)

print("Participant", end="\t")

for hour in range(24):
    print(f"{hour:02d}", end="\t")

print()

for i in range(len(participants_list)):
    print(participants_list[i], end="\t")

    for hour in range(24):
        print(activity_matrix[i][hour], end="\t")

    print()
```

```python
# Find Busiest Hour for Each Participant

print("Busiest Hour for Each Participant")
print("-" * 50)

for i in range(len(participants_list)):
    busiest_hour = 0
    max_messages = activity_matrix[i][0]

    for hour in range(1, 24):
        if activity_matrix[i][hour] > max_messages:
            max_messages = activity_matrix[i][hour]
            busiest_hour = hour

    print(f"{participants_list[i]}: {busiest_hour:02d}:00 - {max_messages} messages")
```

```python
# Overall Busiest Hour

overall_busiest_hour = 0
max_activity = 0

for hour in range(24):
    total = 0

    for i in range(len(participants_list)):
        total += activity_matrix[i][hour]

    if total > max_activity:
        max_activity = total
        overall_busiest_hour = hour

print("Overall Busiest Hour:", f"{overall_busiest_hour:02d}:00")
print("Total Messages:", max_activity)
```

```python
# Activity Heatmap Display

print("\nACTIVITY HEATMAP")
print("=" * 80)

for i in range(len(participants_list)):
    print(f"{participants_list[i]:8}", end=" ")

    for hour in range(24):
        value = activity_matrix[i][hour]

        if value == 0:
            symbol = "."
        elif value <= 2:
            symbol = "░"
        elif value <= 5:
            symbol = "▒"
        else:
            symbol = "█"

        print(symbol, end=" ")

    print()

print("\nHours:")
print("         " + " ".join(f"{h:02d}" for h in range(24)))
```

## Feature 5 — Top Words

This section identifies the most frequently used words in the group chat.
Common filler words and very short words are ignored.

```python
# FEATURE 5: TOP WORDS

word_count = {}

stop_words = {
    "the", "is", "a", "an", "and", "to", "of", "in", "on",
    "for", "this", "that", "it", "i", "you", "we", "he", "she",
    "are", "was", "be", "have", "has", "had", "with", "my", "me",
    "our", "your", "at", "or", "but", "so", "if", "from", "as",
    "just", "not", "will", "can", "all", "do", "dont", "ok"
}

for msg in messages:
    text = msg["message"].lower()

    # Remove common punctuation
    for symbol in ",.!?;:'\"()[]{}":
        text = text.replace(symbol, " ")

    words = text.split()

    for word in words:
        if len(word) > 2 and word not in stop_words:
            if word not in word_count:
                word_count[word] = 0

            word_count[word] += 1

print("Total unique words:", len(word_count))
```

```python
# Sort Words by Frequency

sorted_words = sorted(
    word_count.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Top 20 Most Used Words")
print("-" * 40)

for word, count in sorted_words[:20]:
    print(f"{word:<20} : {count}")
```

```python
# Favourite Words

print("\nFavourite / Most Used Words")
print("-" * 40)

for i, (word, count) in enumerate(sorted_words[:10], start=1):
    print(f"{i}. {word} → {count} times")
```

```python
# Top Words by Participant

print("\nTop Words Used by Each Participant")
print("=" * 50)

for participant in participants_list:
    personal_words = {}

    for msg in messages:
        if msg["sender"] == participant:
            text = msg["message"].lower()

            for symbol in ",.!?;:'\"()[]{}":
                text = text.replace(symbol, " ")

            words = text.split()

            for word in words:
                if len(word) > 2 and word not in stop_words:
                    if word not in personal_words:
                        personal_words[word] = 0

                    personal_words[word] += 1

    personal_sorted = sorted(
        personal_words.items(),
        key=lambda item: item[1],
        reverse=True
    )

    print(f"\n{participant}:")

    for word, count in personal_sorted[:5]:
        print(f"  {word} → {count}")
```

## Feature 6 — Response Speed & Silent Streaks

This section analyses how quickly group members respond to messages and identifies periods when the group stayed silent.

```python
# FEATURE 6: RESPONSE SPEED

from datetime import datetime

response_times = {}

for i in range(1, len(messages)):
    previous_msg = messages[i - 1]
    current_msg = messages[i]

    previous_time = datetime.strptime(
        previous_msg["timestamp"], "%d/%m/%y, %H:%M"
    )

    current_time = datetime.strptime(
        current_msg["timestamp"], "%d/%m/%y, %H:%M"
    )

    # Only consider replies from different participants
    if previous_msg["sender"] != current_msg["sender"]:
        difference = current_time - previous_time
        minutes = difference.total_seconds() / 60

        # Ignore very large gaps
        if minutes <= 60:
            sender = current_msg["sender"]

            if sender not in response_times:
                response_times[sender] = []

            response_times[sender].append(minutes)

print("Response data collected successfully.")
```

```python
# Average Response Time

print("Average Response Time")
print("-" * 50)

average_response = {}

for participant in participants_list:
    if participant in response_times and len(response_times[participant]) > 0:
        total = sum(response_times[participant])
        average = total / len(response_times[participant])
        average_response[participant] = average

        print(f"{participant}: {average:.2f} minutes")
    else:
        average_response[participant] = 0
        print(f"{participant}: No response data")
```

```python
# Fastest Responder

fastest_person = None
fastest_time = None

for participant in participants_list:
    if average_response[participant] > 0:
        if fastest_time is None or average_response[participant] < fastest_time:
            fastest_time = average_response[participant]
            fastest_person = participant

print("Fastest Responder:", fastest_person)
print(f"Average Response Time: {fastest_time:.2f} minutes")
```

```python
# Silent Streaks

silent_streaks = []

for i in range(1, len(messages)):
    previous_time = datetime.strptime(
        messages[i - 1]["timestamp"], "%d/%m/%y, %H:%M"
    )

    current_time = datetime.strptime(
        messages[i]["timestamp"], "%d/%m/%y, %H:%M"
    )

    difference = current_time - previous_time
    minutes = difference.total_seconds() / 60

    if minutes >= 180:
        silent_streaks.append({
            "start": previous_time,
            "end": current_time,
            "minutes": minutes
        })

print("Number of Silent Streaks:", len(silent_streaks))
```

```python
# Longest Silent Streaks

silent_streaks.sort(
    key=lambda item: item["minutes"],
    reverse=True
)

print("Top 5 Longest Silent Streaks")
print("=" * 60)

for streak in silent_streaks[:5]:
    hours = streak["minutes"] / 60

    print(
        f"{streak['start'].strftime('%d-%m-%Y %H:%M')} "
        f"to {streak['end'].strftime('%d-%m-%Y %H:%M')} "
        f"→ {hours:.2f} hours"
    )
```

```python
# Response & Silence Summary

print("=" * 60)
print("        RESPONSE & SILENCE SUMMARY")
print("=" * 60)

print("Fastest Responder:", fastest_person)

if fastest_time is not None:
    print(f"Average Response Time: {fastest_time:.2f} minutes")

print("Total Silent Streaks:", len(silent_streaks))

if len(silent_streaks) > 0:
    longest = silent_streaks[0]["minutes"] / 60
    print(f"Longest Silent Streak: {longest:.2f} hours")
```


#Feature 7 — Participant Archetypes
Classifying participants based on their messaging behavior.

```python
# Calculate average words per message for each participant

average_words = {}

for person in participants_list:

    total_words = 0
    total_messages = 0

    for message in messages:

        if message["sender"] == person:

            text = message["message"]

            # Ignore media messages
            if text == "<Media omitted>":
                continue

            words = text.split()

            total_words += len(words)
            total_messages += 1

    if total_messages > 0:
        average_words[person] = total_words / total_messages
    else:
        average_words[person] = 0

print("Average words per message")
print("=" * 35)

for person in participants_list:
    print(f"{person:<8} {average_words[person]:.2f}")
```

```python

# Calculate messages per active day

from datetime import datetime

person_active_days = {}

message_counts = {}

all_dates = set()

for person in participants_list:
    person_active_days[person] = set()
    message_counts[person] = 0

for message in messages:

    person = message["sender"]

    timestamp = datetime.strptime(
        message["timestamp"], "%d/%m/%y, %H:%M"
    )

    date = timestamp.date()

    all_dates.add(date)
    person_active_days[person].add(date)
    message_counts[person] += 1

messages_per_active_day = {}

for person in participants_list:

    active_days = len(person_active_days[person])

    if active_days > 0:
        messages_per_active_day[person] = (
            message_counts[person] / active_days
        )
    else:
        messages_per_active_day[person] = 0

print("Messages per active day")
print("=" * 35)

for person in participants_list:
    print(
        f"{person:<8} "
        f"{messages_per_active_day[person]:.2f}"
    )
```

```python

# Find the longest consecutive silent streak for each participant

longest_silent_streak = {}

for person in participants_list:

    active_days = person_active_days[person]

    sorted_days = sorted(active_days)

    longest_streak = 0
    current_streak = 0

    if len(sorted_days) > 0:

        for i in range(len(sorted_days) - 1):

            difference = (sorted_days[i + 1] - sorted_days[i]).days

            if difference > 1:
                silent_days = difference - 1

                if silent_days > longest_streak:
                    longest_streak = silent_days

        # Check silence after the last active day
        if len(all_dates) > 0:
            last_active = sorted_days[-1]
            last_date = max(all_dates)

            silent_days = (last_date - last_active).days

            if silent_days > longest_streak:
                longest_streak = silent_days

    longest_silent_streak[person] = longest_streak


print("Longest Silent Streak")
print("=" * 35)

for person in participants_list:
    print(
        f"{person:<8} "
        f"{longest_silent_streak[person]} days"
    )
```

```python
# Calculate caring keyword count

caring_keywords = [
    "okay",
    "safe",
    "eat",
    "sleep",
    "take care",
    "are you",
    "please",
    "reminder",
    "drink water",
    "don't forget"
]

caring_count = {}

for person in participants_list:

    count = 0

    for message in messages:

        if message["sender"] == person:

            text = message["message"].lower()

            for keyword in caring_keywords:

                if keyword in text:
                    count += 1

    caring_count[person] = count


print("Caring Keyword Count")
print("=" * 35)

for person in participants_list:
    print(
        f"{person:<8} "
        f"{caring_count[person]}"
    )
```

```python
# Calculate night activity percentage

night_activity_percentage = {}

for person in participants_list:

    night_messages = 0
    total_messages = 0

    for message in messages:

        if message["sender"] == person:

            total_messages += 1

            timestamp = datetime.strptime(
                message["timestamp"], "%d/%m/%y, %H:%M"
            )

            hour = timestamp.hour

            if hour >= 23 or hour <= 4:
                night_messages += 1

    if total_messages > 0:
        percentage = (night_messages / total_messages) * 100
    else:
        percentage = 0

    night_activity_percentage[person] = percentage


print("Night Activity Percentage")
print("=" * 40)

for person in participants_list:
    print(
        f"{person:<8} "
        f"{night_activity_percentage[person]:.2f}%"
    )
```

```python

# Drama Queen Detection

drama_counts = {}
drama_percentage = {}

for person in participants_list:

    count = 0

    for message in messages:

        if message["sender"] == person:

            text = message["message"]

            if len(text) >= 3 and text.isupper():
                count += 1

            elif text.count("!") >= 2:
                count += 1

    drama_counts[person] = count

    total = message_counts[person]

    if total > 0:
        drama_percentage[person] = (count / total) * 100
    else:
        drama_percentage[person] = 0

print("Drama Queen Percentage")
print("=" * 40)

for person in participants_list:
    print(f"{person:<8} {drama_percentage[person]:.2f}%")
```

```python
# Comedian Detection

comedian_keywords = [
    "lol",
    "lmao",
    "haha",
    "rofl",
    "lmfao"
]

comedian_counts = {}
comedian_percentage = {}

for person in participants_list:

    count = 0

    for message in messages:

        if message["sender"] != person:
            continue

        text = message["message"].lower()

        for keyword in comedian_keywords:
            if keyword in text:
                count += 1
                break

    comedian_counts[person] = count

    total = message_counts[person]

    if total > 0:
        comedian_percentage[person] = (count / total) * 100
    else:
        comedian_percentage[person] = 0

print("Comedian Percentage")
print("=" * 40)

for person in participants_list:
    print(f"{person:<8} {comedian_percentage[person]:.2f}%")
```

```python

# Question Master Detection

question_counts = {}
question_percentage = {}

for person in participants_list:

    count = 0

    for message in messages:

        if message["sender"] == person:

            text = message["message"].strip()

            if text.endswith("?"):
                count += 1

    question_counts[person] = count

    total = message_counts[person]

    if total > 0:
        question_percentage[person] = (count / total) * 100
    else:
        question_percentage[person] = 0

print("Question Master Percentage")
print("=" * 40)

for person in participants_list:
    print(f"{person:<8} {question_percentage[person]:.2f}%")
```

```python

# Average Consecutive Message Burst

burst_values = {}

for person in participants_list:

    bursts = []
    current_burst = 0

    for message in messages:

        if message["sender"] == person:
            current_burst += 1
        else:
            if current_burst > 0:
                bursts.append(current_burst)
            current_burst = 0

    if current_burst > 0:
        bursts.append(current_burst)

    if len(bursts) > 0:
        burst_values[person] = sum(bursts) / len(bursts)
    else:
        burst_values[person] = 0

print("Average Consecutive Message Burst")
print("=" * 45)

for person in participants_list:
    print(f"{person:<8} {burst_values[person]:.2f}")
```

```python
# Calculate Archetype Scores

spammer_score = {}
group_mom_score = {}
night_owl_score = {}
storyteller_score = {}
drama_queen_score = {}
ghost_score = {}
comedian_score = {}
question_master_score = {}

max_caring = max(caring_count.values())
max_comedian = max(comedian_percentage.values())

for person in participants_list:

    # Spammer
    value = burst_values[person]
    spammer_score[person] = value / 3 if value > 3 else 0

    # Group Mom
    if max_caring > 0:
        group_mom_score[person] = caring_count[person] / max_caring
    else:
        group_mom_score[person] = 0

    # Night Owl
    value = night_activity_percentage[person]
    night_owl_score[person] = value / 60 if value > 60 else 0

    # Storyteller
    value = average_words[person]
    storyteller_score[person] = value / 30 if value > 30 else 0

    # Drama Queen
    value = drama_percentage[person]
    drama_queen_score[person] = value / 30 if value > 30 else 0

    # Ghost
    active_days = len(person_active_days[person])
    total_days = len(all_dates)

    if total_days > 0:
        silent_percentage = ((total_days - active_days) / total_days) * 100
    else:
        silent_percentage = 0

    ghost_score[person] = silent_percentage / 60 if silent_percentage > 60 else 0

    # Comedian
    if max_comedian > 0:
        comedian_score[person] = comedian_percentage[person] / max_comedian
    else:
        comedian_score[person] = 0

    # Question Master
    value = question_percentage[person]
    question_master_score[person] = value / 25 if value > 25 else 0

print("Archetype scores calculated successfully!")
```

```python

# Final Archetype Assignment

archetype_scores = {}

for person in participants_list:

    archetype_scores[person] = {
        "The Spammer": spammer_score[person],
        "The Group Mom": group_mom_score[person],
        "The Night Owl": night_owl_score[person],
        "The Storyteller": storyteller_score[person],
        "The Drama Queen": drama_queen_score[person],
        "The Ghost": ghost_score[person],
        "The Comedian": comedian_score[person],
        "The Question Master": question_master_score[person]
    }

final_archetype = {}

for person in participants_list:

    scores = archetype_scores[person]

    best_archetype = max(scores, key=scores.get)

    final_archetype[person] = best_archetype

print("=" * 60)
print("          FINAL PARTICIPANT ARCHETYPES")
print("=" * 60)

for person in participants_list:
    print(f"{person:<8} → {final_archetype[person]}")
```

```python

# Display All Archetype Scores

for person in participants_list:

    print("\n" + person)
    print("-" * 40)

    for archetype in archetype_scores[person]:

        print(
            f"{archetype:<22} "
            f"{archetype_scores[person][archetype]:.2f}"
        )
```
